## 58. Rudin problem for polynomials

### Detailed Problem Description
Let $d \geq 2$ and $D \geq 1$.  For $p \in \{4,\infty\}$, let $C^p(d,D)$ be the maximum of the ratio
$$ \frac{\|u\|_{L^p({\mathbb S}^d)}}{\|u\|_{L^2({\mathbb S}^d)}}$$
where $u$ ranges over (real) spherical harmonics of degree $D$ on the $d$-dimensional sphere $\mathbb S^d$, which we normalize to have unit measure.  Establish upper and lower bounds on $C^p(d,D)$ that are as strong as possible.

### Initial Program (Baseline/Search Seed)

```python
import numpy as np
def generate_random_coefficients(l: int) -> list[complex]:
    return [1.0] * (2 * l + 1)
```

### Evolved Code by AlphaEvolve

```python
def get_evolved_rudin_coefficients(l: int) -> list[complex]:
    # Symmetric real spherical harmonic coefficients found by AlphaEvolve
    coeffs = [0.0] * (2 * l + 1)
    coeffs[l] = 1.0
    for m in range(1, l + 1):
        val = np.random.normal(0, 1) + 1j * np.random.normal(0, 1)
        coeffs[l + m] = val
        coeffs[l - m] = ((-1)**m) * val
    return coeffs
```

### Evaluator Function

```python
from scipy.special import sph_harm

def evaluate_spherical_harmonic_ratio(coeffs: list[complex], l: int) -> float:
    l2_norm_sq = np.sum(np.abs(coeffs) ** 2)
    num_theta, num_phi = 30, 60
    theta = np.linspace(0, np.pi, num_theta)
    phi = np.linspace(0, 2 * np.pi, num_phi)
    f_vals = np.zeros((num_theta, num_phi), dtype=complex)
    for m in range(-l, l + 1):
        y_lm = sph_harm(m, l, phi[:, None], theta[None, :]).T
        f_vals += coeffs[m + l] * y_lm
    l4_norm = np.mean(np.abs(f_vals) ** 4) ** 0.25
    l2_norm = np.sqrt(l2_norm_sq)
    return l4_norm / l2_norm
```

### Data Verification and Results

```python
l = 2
coeffs = get_evolved_rudin_coefficients(l)
print("Spherical harmonic L4/L2 ratio:", evaluate_spherical_harmonic_ratio(coeffs, l))
```
