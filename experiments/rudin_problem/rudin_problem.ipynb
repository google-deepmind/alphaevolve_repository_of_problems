## 58. Rudin problem for polynomials

### Detailed Problem Description
Let $d \geq 2$ and $D \geq 1$.  For $p \in \{4,\infty\}$, let $C^p(d,D)$ be the maximum of the ratio
$$ \frac{\|u\|_{L^p({\mathbb S}^d)}}{\|u\|_{L^2({\mathbb S}^d)}}$$
where $u$ ranges over (real) spherical harmonics of degree $D$ on the $d$-dimensional sphere $\mathbb S^d$, which we normalize to have unit measure.  Establish upper and lower bounds on $C^p(d,D)$ that are as strong as possible.


**Prompt for the search setup**

You are a research mathematician and an optimization specialist specializing in
constructing lists of floats with certain extremal properties.

GOAL:
Your task is to find an algorithm, which for a given degree k (a natural number)
outputs the real coefficients (c_n)_(n=-k)^k of a real linear combination of
spherical harmonics $f_k = sum_(n=-k)^k c_n Y^n_k$ on the
2D unit sphere such that:

1. $f_k$ is real valued.
2. $f_k$ has the lowest possible ratio $\|f_k\|_(L^4) / \|f_k\|_(L^2)$ where
$\|f_k\|_(L^p)$ denotes the standard L^p norm on the round sphere.

Note that the array c_n consists of 2*k + 1 real numbers.

Specifically, the Python function you have to provide is called
get_coefficients(). For example, here is a starting implementation to get you
going:

In other words the get_coefficients functions outputs a list of length 2*k + 1
consisting of the real numbers c_n above.

EVALUATION:
Your proposed method will be scored using the get_score function which you may
call from the context - it takes the generated coefficients as a 1D array and
returns $-\|f_k\|_(L^4) / \|f_k\|_(L^2)$ i.e. the negated ratio we want to
minimize. A higher score (lower ratio) is better. Example usage: for a 1D array
c we call get_score(c).

HINTS:

You may code up any search method you want, and you are allowed to call the

You want the score it gives you to be as high as possible!

Your task is to write a search function that searches for the best list.
Your function will have 300 seconds to run, and after that it has to have
returned the best construction it found. If after 300 seconds it has not

use your time best if you have an outer loop of the form

the "start_time" variable early in your program. You may choose n, the length of

potential for bigger scores, but the search space is larger so they are harder
to find. You'll have to balance it accordingly!

### Initial Program (Baseline/Search Seed)

```python
# @title Initial program

import collections
import time
from typing import Any, Dict, Tuple, Union
import numpy as np
import scipy
import sympy


def get_coefficients(k: int) -> list[float]:
  """Generate coefficients to minimize the norm ratio."""

  variable_name = "suggested_coeffs_" + str(k)
  best_coeffs = []
  best_coeffs.append(np.random.randn())
  for _ in range(1, k + 1):
    best_coeffs.append(np.random.randn())
    best_coeffs.append(np.random.randn())

  if np.random.rand() < 0.5 and variable_name in globals():
    best_coeffs = globals()[variable_name]

  best_coeffs = np.array(best_coeffs)
  curr_coeffs = best_coeffs.copy()
  best_score = get_score(best_coeffs)

  start_time = time.time()

  while time.time() - start_time < 300:
    random_index = np.random.randint(0, len(curr_coeffs))
    curr_coeffs[random_index:] += 1e-1 * np.random.randint(-10, 10)

    curr_score = get_score(curr_coeffs)
    if curr_score > best_score:
      best_score = curr_score
      best_coeffs = curr_coeffs.copy()

  # Wrong shape of coefficients.
  if len(best_coeffs) != 2 * k + 1:
    best_coeffs = np.ones(2 * k + 1)
  # Degenerate coefficients.
  if np.max(np.abs(best_coeffs)) < 1e-3:
    best_coeffs = np.ones(2 * k + 1)
  return best_coeffs
```

### Evaluator Function

```python
# @title Evaluator

import collections
import time
from typing import Any, Dict, Tuple, Union
import numpy as np
import scipy
import sympy


def lm_to_k(l: int, m: int) -> int:
  """Helper to get array index 'k' from an (l, m) pair."""
  return l**2 + l + m


def get_random_spherical_harmonic(
    k: int,
) -> Tuple[np.ndarray, Dict[Tuple[int, int], float]]:
  """Generates a random spherical harmonic of degree k."""
  l_max = k
  n_total = (l_max + 1) ** 2
  c_array = np.zeros(n_total)
  real_coeffs_dict = {}

  random_coeffs = np.random.randn(2 * k + 1)

  for i, m in enumerate(range(-k, k + 1)):
    coeff_val = random_coeffs[i]
    arr_index = lm_to_k(k, m)
    c_array[arr_index] = coeff_val
    real_coeffs_dict[(k, m)] = coeff_val

  return c_array, real_coeffs_dict


def parse_coefficients(
    coeffs: Union[np.ndarray, list[float]],
) -> Tuple[np.ndarray, Dict[Tuple[int, int], float]]:
  """Parses coefficients into the internal array format."""
  k = (len(coeffs) - 1) // 2
  l_max = k
  n_total = (l_max + 1) ** 2
  c_array = np.zeros(n_total)

  for i, m in enumerate(range(-k, k + 1)):
    coeff_val = coeffs[i]
    arr_index = lm_to_k(k, m)
    c_array[arr_index] = coeff_val

  return c_array


def calculate_norm_ratio_analytical(
    real_coeff_array: Union[np.ndarray, list[float]],
) -> float:
  """Computes the exact ratio ||f||_L4 / ||f||_L2 analytically using SymPy."""
  if not isinstance(real_coeff_array, np.ndarray):
    real_coeff_array = np.array(real_coeff_array, dtype=float)
  n = real_coeff_array.shape[0]
  if n == 0:
    return 0.0

  # === Step 1: Parse Input Array ===
  sqrt_n = np.sqrt(n)
  l_max_plus_1 = int(round(sqrt_n))
  if l_max_plus_1**2 != n:
    raise ValueError(f"Input array length N={n} is not a perfect square.")
  real_coeffs_dict = {}
  for k in range(n):
    if not np.isclose(real_coeff_array[k], 0.0):
      l = int(np.floor(np.sqrt(k)))
      m = k - l**2 - l
      real_coeffs_dict[(l, m)] = real_coeff_array[k]
  if not real_coeffs_dict:
    return 0.0

  # === Step 2: Convert to Complex Basis ===
  complex_coeffs_dict = collections.defaultdict(complex)
  l_values = collections.defaultdict(int)
  for l, m in real_coeffs_dict:
    l_values[l] = max(l_values[l], abs(m))
  for l, max_m in l_values.items():
    d_l0 = real_coeffs_dict.get((l, 0), 0.0)
    if not np.isclose(d_l0, 0.0):
      complex_coeffs_dict[(l, 0)] = d_l0
    for m_pos in range(1, max_m + 1):
      d_lm_pos = real_coeffs_dict.get((l, m_pos), 0.0)
      d_lm_neg = real_coeffs_dict.get((l, -m_pos), 0.0)
      if np.isclose(d_lm_pos, 0.0) and np.isclose(d_lm_neg, 0.0):
        continue
      c_lm_pos = (d_lm_pos - 1j * ((-1) ** m_pos) * d_lm_neg) / np.sqrt(2)
      c_lm_neg = ((-1) ** m_pos * d_lm_pos + 1j * d_lm_neg) / np.sqrt(2)
      complex_coeffs_dict[(l, m_pos)] = c_lm_pos
      complex_coeffs_dict[(l, -m_pos)] = c_lm_neg

  # === Step 3: L2-Normalize ===
  l2_norm_sq = sum(abs(c) ** 2 for c in complex_coeffs_dict.values())
  l2_norm = np.sqrt(l2_norm_sq)
  if np.isclose(l2_norm, 0):
    return 0.0
  normalized_complex_coeffs = {
      k: v / l2_norm for k, v in complex_coeffs_dict.items()
  }
  coeff_items = list(normalized_complex_coeffs.items())

  # === Step 4: Get Coefficients of |g|^2 ===
  coeffs_of_g_squared = collections.defaultdict(complex)
  for (l1, m1), c1 in coeff_items:
    for (l2, m2), c2 in coeff_items:
      m_val = m1 - m2
      l_start, l_end = abs(l1 - l2), l1 + l2
      if (l1 + l2 + l_start) % 2 != 0:
        l_start += 1
      for l in range(l_start, l_end + 1, 2):

        w3j_1_sym = sympy.physics.wigner.wigner_3j(l1, l2, l, 0, 0, 0)
        w3j_1 = float(w3j_1_sym.evalf())
        if np.isclose(w3j_1, 0.0):
          continue

        w3j_2_sym = sympy.physics.wigner.wigner_3j(l1, l2, l, m1, -m2, -m_val)
        w3j_2 = float(w3j_2_sym.evalf())
        if np.isclose(w3j_2, 0.0):
          continue

        prefactor = np.sqrt(
            (2 * l1 + 1) * (2 * l2 + 1) * (2 * l + 1) / (4 * np.pi)
        )
        gaunt_val = prefactor * w3j_1 * w3j_2

        term = c1 * np.conj(c2) * ((-1) ** m2) * gaunt_val
        coeffs_of_g_squared[(l, m_val)] += term

  # === Step 5: Calculate Final Norm ===
  l4_norm_fourth_power = 0.0
  for (l, m), b_lm in coeffs_of_g_squared.items():
    b_l_neg_m = coeffs_of_g_squared.get((l, -m), 0.0)
    l4_norm_fourth_power += (-1) ** m * b_lm * b_l_neg_m

  # === Step 6: Return Result ===
  return np.power(np.real(l4_norm_fourth_power), 0.25)


def get_score(coeffs: Union[np.ndarray, list[float]]) -> float:
  # Degenerate coefficients.
  score = -calculate_norm_ratio_analytical(parse_coefficients(coeffs))
  return score


def evaluate(
    degree: int = 10,
) -> tuple[dict[str, float], dict[str, str]]:
  """Returns the numerical bound for the polygons if valid, or 0 if invalid."""
  result = {}
  feedback = {}

  # Start construction evaluation
  time_start = time.time()
  xs = get_coefficients(degree)
  if len(xs) != 2 * degree + 1:
    feedback["error_message"] = "Wrong shape of coefficients."
    result["score_max"] = -10000.0
    return result, feedback

  if np.max(np.abs(xs)) < 1e-3:
    feedback["error_message"] = "Too small coefficients."
    result["score_max"] = -10000.0
    return result, feedback

  if np.max(np.abs(xs)) > 1e5:
    feedback["error_message"] = "Too large coefficient appearing."
    result["score_max"] = -10000.0
    return result, feedback

  feedback["degree"] = str(degree)
  feedback["suggested_coeffs"] = xs.copy()

  print(f"Degree: {degree}, Suggested coefficients: {xs}")

  score = get_score(xs)
  result["score_max"] = score
  feedback["ratio_value"] = score

  print(f"Time taken for full evaluation: {time.time() - time_start:.2f}s")

  return result, feedback
```
