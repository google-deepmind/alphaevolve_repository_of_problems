## 51. Erdős–Gyárfás conjecture

### Detailed Problem Description
Let $G$ be a finite graph with minimum degree at least $3$.  Must $G$ contain a cycle of length $2^k$ for some $k \geq 2$?

### Initial Program (Baseline/Search Seed)

```python
import networkx as nx
def get_random_regular_graph(n: int) -> nx.Graph:
    return nx.random_regular_graph(3, n)
```

### Evolved Code by AlphaEvolve

```python
def evolved_edge_removal_heuristic(G: nx.Graph) -> nx.Graph:
    # Evolved heuristic: greedily remove edges between vertices of degree > 3
    refined_G = G.copy()
    for u, v in list(refined_G.edges()):
        if refined_G.degree(u) > 3 and refined_G.degree(v) > 3:
            refined_G.remove_edge(u, v)
    return refined_G
```

### Evaluator Function

```python
import networkx as nx

def evaluate_gyarfas_graph(G: nx.Graph) -> float:
    penalty = 0
    for node in G.nodes():
        deg = G.degree(node)
        if deg < 3: penalty += (3 - deg) * 100
    cycles = nx.cycle_basis(G)
    power_2_cycles = 0
    for cycle in cycles:
        length = len(cycle)
        if length >= 4 and (length & (length - 1)) == 0:
            power_2_cycles += 1
    return - (penalty + power_2_cycles)
```

### Data Verification and Results

```python
G = nx.complete_graph(4)
print("Score for K_4 (degree 3, has a cycle of length 4):", evaluate_gyarfas_graph(G))
```
