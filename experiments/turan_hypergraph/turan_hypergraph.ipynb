### Evolved Code by AlphaEvolve

```python
def get_optimal_turan_hypergraph():
    # Recovered 5/9 construction
    weights = [1.0/3.0, 1.0/3.0, 1.0/3.0]
    edges = [(0, 1, 2), (0, 0, 1), (1, 1, 2), (2, 2, 0)]
    return weights, edges
```

### Evaluator Function

```python
def compute_turan_density(weights: list[float], edges: list[tuple]) -> float:
    w = [x / sum(weights) for x in weights]
    density = 0.0
    for edge in edges:
        a, b, c = edge
        if len(set(edge)) == 3:
            density += 6 * w[a] * w[b] * w[c]
        elif len(set(edge)) == 2:
            dup = a if edge.count(a) == 2 else (b if edge.count(b) == 2 else c)
            other = list(set(edge) - {dup})[0]
            density += 3 * w[dup] * w[dup] * w[other]
        elif len(set(edge)) == 1:
            density += w[a]**3
    return density
```
