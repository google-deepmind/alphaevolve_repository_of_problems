## 52. Erdős squarefree problem

### Detailed Problem Description
For any natural number $N$,
let $C(N)$ denote the largest cardinality of a subset $A$ of $\{1,\dots,N\}$ with the property that $ab+1$ is not square-free for all $a,b \in A$.  Establish upper and lower bounds for
$C(N)$ that are as strong as possible.

### Evolved Code by AlphaEvolve

```python
def get_optimal_squarefree_subset(N: int) -> list[int]:
    # Recovered 7 mod 25 residue class construction
    return [x for x in range(1, N + 1) if x % 25 == 7]
```

### Evaluator Function

```python
def is_square_free(n: int) -> bool:
    if n % 4 == 0: return False
    for i in range(3, int(n**0.5) + 1, 2):
        if n % (i * i) == 0: return False
    return True

def evaluate_squarefree_subset(N: int, subset: list[int]) -> float:
    violations = 0
    n = len(subset)
    for i in range(n):
        for j in range(i+1, n):
            if is_square_free(subset[i] * subset[j] + 1):
                violations += 1
    return len(subset) / N - violations
```

### Data Verification and Results

```python
N = 500
subset = get_optimal_squarefree_subset(N)
print("Score for 7 mod 25 subset:", evaluate_squarefree_subset(N, subset))
```
