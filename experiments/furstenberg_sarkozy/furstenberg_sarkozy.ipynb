## 32. Furstenberg-Sárközy Problem

### Detailed Problem Description
If $k, m \geq 2$ and $N \geq 1$, let $C(k,N)$ (resp. $C(k,\Z/M\Z)$) denote the size of the largest subset of $\{1,\dots,N\}$
that does not contain any two elements that differ by a perfect $k^{\mathrm{th}}$ power. Establish upper and lower bounds for $C(k,N)$ and $C(k,\Z/M\Z)$ that are as strong as possible.

### Evolved Code by AlphaEvolve

```python
def get_optimal_sarkozy_subset(m: int, power: int) -> list[int]:
    # Reproducing optimal modular cliques found by AlphaEvolve
    if m == 205 and power == 2:
        return [0, 5, 27, 38, 57, 76, 104, 115, 134, 145, 167, 178]
    elif m == 91 and power == 3:
        return [0, 4, 13, 20, 24, 30, 40, 51, 61, 67, 71, 78, 87, 88]
    return []
```

### Evaluator Function

```python
def check_sarkozy_subset(m: int, subset: list[int], power: int = 2) -> bool:
    n = len(subset)
    powers = set(pow(x, power, m) for x in range(1, m))
    for i in range(n):
        for j in range(i+1, n):
            diff = (subset[i] - subset[j]) % m
            diff2 = (subset[j] - subset[i]) % m
            if diff in powers or diff2 in powers:
                return False
    return True
```

### Data Verification and Results
